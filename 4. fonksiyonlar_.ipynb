## **4.1 Fonksiyon Yazma**

$$ f(x) = x^2 + 2x + 1 $$

Benzer şekilde, Python'da bir fonksiyon tanımlamak için `def` anahtar kelimesi kullanılır.

## **4.2 Fonksiyon Argümanları ve Yerel Değişkenler**

Bir fonksiyon, birden fazla **argüman (parametre)** alabilir ve işleyebilir. Fonksiyon içinde tanımlanan değişkenler, **yerel değişkenlerdir (local variables)** ve sadece o fonksiyon içinde geçerlidir.

Fonksiyona birden fazla parametre gönderilebilir.
```python
def fonksiyon_adi(parametre1, parametre2, ...):
```

## **4.4 Koşullu Dallanma (if-else)**

Koşullu ifadeler, belirli bir **şartın doğru olup olmadığına** bağlı olarak farklı işlemler yapmamızı sağlar.

````python
def pozitif_mi(x):
    if x > 0:
        return 'Pozitif'
    elif x == 0:
        return 'Sıfır'
    else:
        return 'Negatif'

print(pozitif_mi(10))  # Pozitif
print(pozitif_mi(0))   # Sıfır
print(pozitif_mi(5))  # Negatif
````

## **4.5 Fonksiyonları Argüman Olarak Kullanma**

Python'da bir fonksiyon, başka bir fonksiyona **argüman olarak** verilebilir.

````python
def iki_kat(x):
    return x * 2

def uygulama(f, x):
    return f(x)


print(uygulama(iki_kat, 5))
````

````python
import math as m
def kare(x):
    return x ** 2

def hipotenus(a, b):
    return m.sqrt(kare(a) + kare(b))

print(hipotenus(3, 4))
````

##**4.6. Fonksiyon İçinde Fonksiyon**

Bir fonksiyonun içinde başka bir fonksiyon tanımlayabiliriz.
Bu yapı genellikle kapsülleme (encapsulation) veya özel yardımcı fonksiyon gerektiğinde kullanılır.

Örnek:


````python
def dis_fonksiyon(x):
    def ic_fonksiyon(y):
        return y ** 2
    return ic_fonksiyon(x) + 10

print(dis_fonksiyon(3))
````

## **4.7 Python ile Denklem Çözme**
Python ile denklem çözümlerini yapabiliriz. Çözüm için sayısal yöntemleri kendimiz tanımlayabileceğimiz gibi ``scipy`` gibi hazır paketleri de kullanabiliriz.

Örnek olarak newton yöntemi ile başlayalım.

Newton yöntemi, bir fonksiyonun kökünü (yani ( f(x) = 0 ) denkleminin çözümünü) **türev tabanlı** olarak bulmak için kullanılan sayısal bir yöntemdir.
Yaklaşım, fonksiyonun **teğet doğrularını** kullanarak köke adım adım yaklaşır.

$$
x_{n+1} = x_n - \frac{f(x_n)}{f'(x_n)}
$$


````python
from math import cos

def f(x):
    return cos(x) - x

def newton_yontemi(f, x0, tol=1e-6):
    h = 1e-5  # türev için küçük fark
    while abs(f(x0)) > tol:
        # Sayısal türev: f'(x) ≈ (f(x+h) - f(x)) / h
        x0 = x0 - f(x0) / ((f(x0 + h) - f(x0)) / h)
        print(x0,abs(f(x0))-tol)
    return x0

kok = newton_yontemi(f, 0.5)
print(f'Kök: {kok}')
````

Bu örnekte
$( f(x) = \cos(x) - x )$ denkleminin kökü yaklaşık olarak:

$$
x \approx 0.739085
$$

olarak bulunur. Bu, `x = cos(x)` denkleminin sabit noktasını temsil eder.



Python’da denklemleri çözmek için iki temel yaklaşım vardır:

1. **Sayısal (Numerical) yöntemlerle çözüm** – `sympy` kullanmadan, doğrudan fonksiyon ve döngülerle
2. **Sembolik (Symbolic) çözüm** – `sympy` kütüphanesiyle bilinmeyenli denklemleri çözmek

### 🧮 1. Fonksiyon Tanımlayarak Denklemi Temsil Etme

Bir denklemi önce **fonksiyon olarak** tanımlarız:


````python
def f(x):
    return x**2 - 4*x + 3
````


### 🔍 2. Sayısal Yöntemle (Kök Arama)

Kök bulmak için `scipy` kullanabiliriz:

````python
from scipy.optimize import fsolve

def f(x):
    return x**2 - 4*x + 3

kok = fsolve(f, 0)   # 0 tahminiyle başlar
print(kok)
````

### 🧠 3. Sembolik (cebirsel) Yöntem: `sympy`

``sympy`` kütüphanesi ile bir değişken tanımlayıp cebirsel işlemler yapabiliriz.

````python
import sympy as sp

x = sp.Symbol('x')#sembolü tanımladık
denklem = x**2 - 4*x + 3

kokler = sp.solve(denklem, x)
print(kokler)
````

### ⚙️ 4. Birden Fazla Denklem (Denklem Sistemi)
``sympy`` ile birden çok denklemi çözebiliriz.



````python
import sympy as sp

x, y = sp.symbols('x y')
denklem1 = sp.Eq(x + y, 5)
denklem2 = sp.Eq(x - y, 1)

sonuc = sp.solve((denklem1, denklem2), (x, y))
print(sonuc)
````

### 🧩 5. Fonksiyonla Çözümü Genelleştirmek

Kendimiz küçük bir “denklem çözücü fonksiyon” da yazabiliriz:


````python
import sympy as sp

def denklem_coz(ifade, degisken='x'):
    x = sp.Symbol(degisken)
    cozum = sp.solve(ifade, x)
    return cozum

print(denklem_coz('x**2 - 5*x + 6'))
````

## **4.8 Lambda (Anonim) Fonksiyonlar**

Python’da **lambda fonksiyonları**, tek satırlık kısa fonksiyonlar yazmak için kullanılır.
Bu fonksiyonlar **isimsizdir (anonimdir)** ve genellikle kısa süreli işlemlerde kullanılır.
Yapısı;

```python
lambda parametreler : ifade
```
biçimidedir.Buradaki `ifade`, fonksiyonun **geri döndüreceği değeri** belirtir.
`return` anahtar sözcüğü kullanılmaz.
Basit Örnek:



````python
kare = lambda x: x ** 2
print(kare(5))
````

### 🔹 **Birden Fazla Parametre**



````python
topla = lambda a, b: a + b
print(topla(3, 7))
````

### 🔹 **Lambda ve map() ile Liste İşlemleri**



````python
sayilar = [1, 2, 3, 4, 5]
kareler = list(map(lambda x: x**2, sayilar))
print(kareler)
````


### 🔹 **Lambda ve filter() ile Koşullu Seçim**


````python
sayilar = [1, 2, 3, 4, 5, 6]
ciftler = list(filter(lambda x: x % 2 == 0, sayilar))
print(ciftler)
````

### 🔹 **Lambda ve sorted() ile Sıralama**





````python
veri = [(1, 3), (2, 2), (3, 1)]
sirali = sorted(veri, key=lambda x: x[1])
print(sirali)
````


### 💡 **Ne Zaman Kullanılır?**

* Tek satırlık, basit işlemler gerektiğinde
* `map()`, `filter()`, `sorted()` gibi fonksiyonlarla kısa ifade tanımlamak için
* Kodun okunabilirliğini artırmak için (gereksiz fonksiyon tanımlamadan)

## ⚖️ **4.9 Lambda ve def Arasındaki Farklar**

Python’da hem `def` hem de `lambda` ifadeleri **fonksiyon tanımlamak** için kullanılır.
Ancak amaçları ve kullanım biçimleri farklıdır.

---





### 🧩 **1. Yapı (Sözdizimi) Farkı**

| Özellik            | `def` ile Fonksiyon           | `lambda` Fonksiyonu                       |
| ------------------ | ----------------------------- | ----------------------------------------- |
| **Tanım Şekli**    | Birden fazla satır içerebilir | Tek satırda yazılır                       |
| **Anahtar Sözcük** | `def`                         | `lambda`                                  |
| **İsim**           | Genellikle isimlidir          | İsimsiz (anonim)                          |
| **return**         | Kullanılır                    | Kullanılmaz (ifade sonucu otomatik döner) |

---

### 🔹 **Örnek 1: def ile**

```python
def kare(x):
    return x ** 2
```

### 🔹 **Örnek 2: lambda ile**

```python
kare = lambda x: x ** 2
```

İkisi de aynı sonucu verir:

```python
print(kare(5))  # 25
```


---

### 🧮 **2. Kapsam ve Karmaşıklık**

* `def`: Birden fazla ifade, koşul veya döngü içerebilir.
* `lambda`: Yalnızca **tek bir ifade** (tek satır) içerebilir.

**Örnek:**

```python
# def ile
def kontrol(x):
    if x > 0:
        return "Pozitif"
    else:
        return "Negatif veya sıfır"

# lambda ile (tek satırda)
kontrol = lambda x: "Pozitif" if x > 0 else "Negatif veya sıfır"
```

---

### 🧠 **3. Nerede Kullanılırlar?**

| Kullanım Durumu                                | Tercih     |
| ---------------------------------------------- | ---------- |
| Tek satırlık basit işlem                       | ✅ `lambda` |
| Birden fazla adım veya karmaşık mantık         | ✅ `def`    |
| Sık kullanılacak fonksiyonlar                  | ✅ `def`    |
| Kısa süreli, bir kez kullanılacak fonksiyonlar | ✅ `lambda` |

---



### 🧩 **4. Örnek: Sıralama İşlemi**

```python
veriler = [(1, 3), (2, 2), (3, 1)]

# lambda ile
sirali = sorted(veriler, key=lambda x: x[1])

# def ile
def siralama_kriteri(x):
    return x[1]
sirali2 = sorted(veriler, key=siralama_kriteri)

print(sirali)
print(sirali2)
```

---

### 🎯 **Özet**

| Kriter            | `def`                | `lambda`          |
| ----------------- | -------------------- | ----------------- |
| Satır sayısı      | Çok satırlı olabilir | Tek satır         |
| return            | Kullanılır           | Gerekmez          |
| İsmi              | Genellikle vardır    | Genellikle yoktur |
| Karmaşık işlem    | Uygun                | Uygun değil       |
| Kısa geçici işlem | Gereksiz             | İdeal ✅           |
