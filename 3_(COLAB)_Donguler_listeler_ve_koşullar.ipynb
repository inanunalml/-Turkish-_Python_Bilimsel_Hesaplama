## 3.1 Döngüler ile Tekrarlayan İşlemleri Otomatikleştirme

### 🔁 Döngü Nedir?

**Döngü (loop)**, bir programda belirli bir işlemin **tekrar tekrar yapılmasını** sağlayan yapıdır.
Aynı işlemi birçok kez elle yazmak yerine, döngüler sayesinde bu işlem **otomatik olarak** yinelenir.

#### 🔹 Python’da Temel Döngü Türleri
Temel olarak iki döngü türü bulunur. Bunlar ``for`` ve ``while`` döngüleridir.

| Döngü Türü      | Açıklama                                                                        | Kullanım                         |
| --------------- | ------------------------------------------------------------------------------- | ----------------------------- |
| `for` döngüsü   | Belirli bir sayı kadar veya bir liste üzerindeki elemanlar boyunca tekrar eder. | `for i in range(5): print(i)` |
| `while` döngüsü | Koşul **True** olduğu sürece çalışır.                                           | `while x < 10: x += 1`        |



`````python
#Örnek
for i in range(5):
  print("Merhaba Python!")
`````


### 🌍 **Diğer Dillerde `for` Döngüsü – Kısa Karşılaştırma**

| Dil            | Örnek Kod                                        | Açıklama                                     |
| -------------- | ------------------------------------------------ | -------------------------------------------- |
| **Python**     | `for i in range(5): print(i)`                    | Otomatik artan, sade sözdizimi               |
| **C**          | `for(int i=0;i<5;i++){ printf("%d",i); }`        | Üç bölümden oluşur: başlangıç–koşul–arttırma |
| **Java**       | `for(int i=0;i<5;i++){ System.out.println(i); }` | C’ye benzer, tip zorunlu                     |
| **JavaScript** | `for(let i=0;i<5;i++){ console.log(i); }`        | Web ortamında kullanılır                     |

📘 **Özet:**
Python’da `for` döngüsü **koleksiyon temelli**,
diğer dillerde ise genellikle **sayaç temellidir**.




`while` döngüsü, **belirli bir koşul `True` olduğu sürece** çalışmaya devam eder.
Koşul `False` olduğunda döngü sona erer.

```python
i = 0
while i < 5:
    print(i)
    i += 1
```

📤 **Çıktı:**

```
0  
1  
2  
3  
4
```

📘 **Not:**
Eğer koşul hiç `False` olmazsa döngü **sonsuz döngüye** girer — bu yüzden sayaç veya çıkış koşulu eklenmelidir.


`````python
#while için başka örnek (faiz hesabı)

P = 100  # Başlangıç miktarı (TL)
r = 5.0  # Faiz oranı (%)
n = 0  # Yıl sayacı
while n <= 10:
    A = P * (1 + r / 100) ** n
    print(f'Yıl: {n}, Miktar: {A:.2f} TL')
    n += 1
`````

### 🔍 `for` Döngüsünde Dikkat Edilmesi Gereken Hususlar

`for` döngüsü Python’da **tekrarlayan işlemleri** kolaylaştırır, ancak doğru çalışması için bazı noktalara dikkat edilmelidir:

---

#### 🔹 1. **Döngü Değişkeninin Kontrolü**

* Döngüde kullanılan değişken (örneğin `i`) döngü bittikten sonra da **tanımlı** kalır.
  Bu nedenle başka bir yerde aynı ismi kullanıyorsan, değeri değişmiş olabilir.




`````python
  for i in range(5):
      print(i)
  print("Döngü Sonunda i =", i)
`````

#### 🔹 2. **`range()` Fonksiyonunun Mantığı**

* `range(a, b)` ifadesi **`a` dahil**, **`b` hariç** olacak şekilde sayıları üretir.
---

`````python
for i in range(1, 5): #range(3)=range(0,3)
      print(i)
`````

#### 🔹 3. **İç İçe Döngüler (Nested Loops)**

* İç döngülerin her biri dış döngüye bağlı çalışır.
  Bu nedenle işlem sayısı **katlanarak artabilir**.


`````python
for i in range(3):
    for j in range(2):
          print(i, j)
`````


#### 🔹 4. **Döngülerin Sonu**

* Python’da `for` döngüleri sonunda **else** bloğu da kullanılabilir.
  Döngü **normal bittiğinde** (break olmadan) çalışır:


`````python
for i in range(3):
      print(i)
else:
      print("Döngü tamamlandı.")
`````

### ⏹️ **`while` Döngüsünden Çıkma**

`while` döngüsünde çıkış yapmak için genellikle iki özel komut kullanılır:

---

#### 🔹 **1. `break` → Döngüyü tamamen bitirir**

Koşul hâlâ `True` olsa bile döngüyü **hemen durdurur**.

```python
i = 0
while i < 10:
    if i == 5:
        break
    print(i)
    i += 1
```

📤 **Çıktı:** `0 1 2 3 4`

---

#### 🔹 **2. `continue` → Sadece o adımı atlar**

O anki turu **atlayıp** döngünün başına döner.

```python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

📤 **Çıktı:** `1 2 4 5`
(3 atlandı)

---

📘 **Özet:**

* `break` → Döngü **tamamen sona erer**
* `continue` → Döngü **devam eder ama o adımı atlar**


## 3.2 ⚙️ Boolean İfadeler (Mantıksal İfadeler)

**Boolean ifadeler**, `True` veya `False` değerleri alabilen mantıksal ifadeleridir. Bu ifadeler genellikle karşılaştırmalarda ve koşullu ifadelerde kullanılır.


**Boolean ifadeler**, bir koşulun **doğru (True)** veya **yanlış (False)** olup olmadığını belirler.
Python’da **koşul ifadelerinde** ve **döngülerde** çok sık kullanılır.

---

#### 🔹 1. **Temel Boolean Değerleri**

* Python’da yalnızca iki mantıksal değer vardır:

  ```python
  True
  False
  ```

---

#### 🔹 2. **Karşılaştırma Operatörleri**

| Operatör | Anlamı          | Örnek    | Sonuç  |
| -------- | --------------- | -------- | ------ |
| `==`     | Eşit mi         | `5 == 5` | `True` |
| `!=`     | Eşit değil mi   | `4 != 3` | `True` |
| `>`      | Büyüktür        | `7 > 2`  | `True` |
| `<`      | Küçüktür        | `2 < 8`  | `True` |
| `>=`     | Büyük veya eşit | `6 >= 6` | `True` |
| `<=`     | Küçük veya eşit | `3 <= 5` | `True` |

---

#### 🔹 3. **Mantıksal Operatörler**

| Operatör | Anlamı                       | Örnek              | Sonuç             |
| -------- | ---------------------------- | ------------------ | ----------------- |
| `and`    | İki koşul da **True** olmalı | `x > 0 and x < 10` | `True` (x=5 ise)  |
| `or`     | En az biri **True** olmalı   | `x < 0 or x > 10`  | `False` (x=5 ise) |
| `not`    | Sonucu tersine çevirir       | `not (x > 0)`      | `False` (x=5 ise) |

---


`````python
t = 40
print(t == 40)  # True
print(t != 40)  # False
print(t > 30 and t < 50)  # True
`````

#### 🔹 4. **Boolean İfadelerin Döngülerde Kullanımı**

Boolean ifadeler genellikle `while` döngülerinde veya `if` koşullarında kullanılır:


`````python
x = 0
while x < 5:
  print(x)
  x += 1
`````


#### 🔹 5. **Boolean Dönüşümü**

Python’da bazı değerler otomatik olarak Boolean’a çevrilebilir:

| Değer                         | Boolean Karşılığı |
| ----------------------------- | ----------------- |
| `0`, `""`, `[]`, `{}`, `None` | `False`           |
| Diğer tüm değerler            | `True`            |


`````python
bool(1)
`````

`````python

bool("Python")
`````

## **3.3 Listeler ile Veri Depolama**

Listeler, birden fazla değeri tek bir değişkende saklamak için kullanılan veri yapılarıdır. Python'da listeler köşeli parantez `[]` ile tanımlanır.

**Liste (list)**, Python’da **birden fazla değeri tek bir değişkende saklamaya** yarayan veri tipidir.
Listeler, sıralı (ordered), değiştirilebilir (mutable) ve farklı veri tiplerini bir arada tutabilir.

---



#### 🔹 1. **Liste Tanımlama**

Liste köşeli parantezlerle (`[]`) oluşturulur:

`````python
meyveler = ["elma", "muz", "çilek"]
sayilar = [1, 2, 3, 4, 5]
karisik = [1, "merhaba", True, 3.14]
meyveler
`````


#### 🔹 2. **Liste Elemanlarına Erişim**

* Elemanlara **indeks numarası** ile ulaşılır.
* İndeksler **0’dan başlar**.

`````python
print(meyveler[0])
print(meyveler[2])
print(meyveler[-2])
`````

> Negatif indeksler sondan başa doğru sayar:
> `meyveler[-1]` → "çilek"


#### 🔹 3. **Eleman Ekleme ve Silme**

Listelere sonradan eleman eklenebilir ve bazı elemanlar silinebilir.

`````python
meyveler.append("portakal")     # Sona ekler
meyveler.insert(1, "kiraz")     # 1. indekse ekler
meyveler.remove("elma")         # Belirtilen elemanı siler
meyveler.pop(5)
meyveler
`````

#### 🔹 4. **Liste Uzunluğu**
Listedeki elemanların sayısı aslında listenin uzunluğunu verir. ``len`` metodu elde edilir.  




`````python
print(len(meyveler))
`````

#### 🔹 5. **Dilimleme (Slicing)**
Bire liste alt listelere ayrılabilir.


`````python
sayilar = [0, 1, 2, 3, 4, 5, 6] #yeni bir liste
print(sayilar[2:5])
print(sayilar[:3])
print(sayilar[1:])
print(sayilar[:])
print(sayilar[::2])
print(sayilar[::-1])
print(sayilar[2::-1])
print(sayilar[4:2:-1])
`````

#### 🔹 6. **Listeler Üzerinde Döngü**
Döngüleri liste elemanları üzerinde çalıştırabiliriz. Bu bize listelerdeki elemanlarla işlemler yapma konusunda avantaj sağlar.

`````python
for meyve in meyveler:
    print(meyve)
`````



#### 🔹 7. **Liste Metotları** (Özet)

| Metot           | Açıklama                            | Örnek                  |
| --------------- | ----------------------------------- | ---------------------- |
| `.append(x)`    | Sona eleman ekler                   | `liste.append(5)`      |
| `.insert(i, x)` | Belirtilen konuma ekler             | `liste.insert(2, "a")` |
| `.remove(x)`    | Elemanı siler                       | `liste.remove("a")`    |
| `.pop(i)`       | Belirtilen indeksteki elemanı siler | `liste.pop(1)`         |
| `.sort()`       | Sıralar                             | `liste.sort()`         |
| `.reverse()`    | Ters çevirir                        | `liste.reverse()`      |
| `.count(x)`     | Eleman sayısını döner               | `liste.count("elma")`  |
| `.index(x)`     | Elemanın indeksini döner            | `liste.index("elma")`  |

---

#### 🔹 8. **Liste Kapsamı (List Comprehension)**
List Comprehension Python'da en kullanışlı yöntemlerden biridir. for döngüsü doğrudan liste içinde çalışır ve böylece liste elemanları oluşturulur.
> Bu yöntemin avantaj ve dezavantajlarını araştırın...


`````python
kareler = [x**2 for x in range(1, 6)]
print(kareler)
`````

## **3.4 İç İçe Listeler (Nested Lists)**

Python’da **liste içinde liste** tanımlayarak **çok boyutlu veri yapıları** oluşturabiliriz.
Bu yapıya **iç içe liste (nested list)** denir.

---



#### 🔹 1. **İç İçe Liste Tanımlama**

Bir listenin elemanları başka listeler olabilir. Aşağıdaki matrisi inceleyelim.

`````python
matris = [[1, 2, 3],[4, 5, 6],[7, 8, 9]]
matris
`````


📌 Burada `matris`, 3 satır ve 3 sütundan oluşan 2 boyutlu bir listedir.

#### 🔹 2. **Elemanlara Erişim**

İlk indeks **satırı**, ikinci indeks **sütunu** belirtir:


`````python
print(matris[0])
print(matris[1][2])
print(matris[2][0])
`````

#### 🔹 3. **İç İçe Döngü ile Liste Elemanlarını Gezme**

İç içe listelerde tüm elemanlara ulaşmak için **iç içe döngüler** kullanılır.

`````python
for satir in matris:
    for eleman in satir:
        print(eleman, end=" ")
`````


#### 🔹 4. **Eleman Güncelleme**

İç listedeki belirli bir elemanı değiştirmek mümkündür:


`````python
matris[1][1] = 99
print(matris)
`````



#### 🔹 5. **Karmaşık Örnek**
Biraz karmaşık bir örneğe bakalım:


`````python
ogrenciler = [["Ali", [80, 90, 85]],
    ["Ayşe", [75, 95, 88]],
    ["Mehmet", [60, 70, 65]]
]
# Ayşe'nin 2. notu:
print(ogrenciler[1][1][1])
`````


#### 🔹 6. **List Comprehension ile İç İçe Liste**
List comprehension ile iç içe listeler üretebiliriz.  

`````python
kare_matris = [[i*j for j in range(1, 4)] for i in range(1, 4)]
print(kare_matris)
`````

## **3.6 Demetler (Tuples)**

**Tuple**, Python’da **birden fazla veriyi tek bir değişkende saklamak** için kullanılan,
ancak **değiştirilemeyen (immutable)** bir veri yapısıdır.
Listelere benzer ama **elemanları sonradan değiştirilemez**.

---

#### 🔹 1. **Tuple Tanımlama**

Tuple, **parantez** `( )` kullanılarak tanımlanır.

`````python
renkler = ("kırmızı", "yeşil", "mavi")
sayilar = (1, 2, 3, 4)
karisik = (1, "Python", True, 3.14)
`````

Tek elemanlı tuple tanımlarken **sonuna virgül** koymak gerekir:


`````python
tek = (5,)    # Tuple
degil = (5)   # Sadece int
print(tek)
print(degil)
type(degil)
`````

#### 🔹 2. **Elemanlara Erişim**

Listelerde olduğu gibi **indeksleme** yapılabilir.


`````python
print(renkler[0])   # kırmızı
print(renkler[-1])  # mavi
`````

#### 🔹 3. **Dilimleme (Slicing)**
Listelerdeki gibi yapılır.

`````python
sayilar = (10, 20, 30, 40, 50)
print(sayilar[1:4])   # (20, 30, 40)
`````

#### 🔹 4. **Değiştirilemezlik (Immutable)**

Tuple’lar oluşturulduktan sonra değiştirilemez.


`````python
renkler[1] = "sarı"
`````


Ancak yeni bir tuple oluşturulabilir.

`````python
renkler = renkler + ("sarı",)
print(renkler)
`````

#### 🔹 5. **İterasyon (Döngü ile Gezme)**

Listelerde olduğu gibi döngü kullanılarak tuple da gezilebilir.


`````python
for renk in renkler:
    print(renk)
`````

#### 🔹 6. **Tuple Fonksiyonları**

| Fonksiyon  | Açıklama                    | Örnek               |
| ---------- | --------------------------- | ------------------- |
| `len()`    | Eleman sayısını verir       | `len(renkler)`      |
| `count(x)` | x değerinden kaç tane var   | `sayilar.count(20)` |
| `index(x)` | x değerinin indeksini döner | `sayilar.index(30)` |

---

### 🎁 **7. Tuple Paketleme (Packing) ve Açma (Unpacking)**

**Paketleme (packing)**, birden fazla değeri tek bir tuple içinde toplamaktır.
**Açma (unpacking)** ise bu tuple’daki değerleri ayrı değişkenlere atamaktır.

`````python
veri = ("Ali", 25, "Mühendis")   # Paketleme
(ad, yas, meslek) = veri         # Açma
print(ad, yas, meslek)
`````


#### 🔹 8. **İç İçe Tuple (Nested Tuple)**
Demetleri iç içe yazabiliriz.


`````python
koordinatlar = ((0, 0), (1, 2), (3, 5))
print(koordinatlar[1][0])  # 1
`````


#### 🔹 9. **Tuple ve Liste Arasında Dönüşüm**
Bir liste bir tuple ve bir tuple bir liste yapılabilir.

`````python
liste = [1, 2, 3]
demet = tuple(liste)
print(demet)
`````

`````python
yeni_liste = list(demet)
print(yeni_liste)
`````

## **3.7. Matematiksel Örnekler**
Şimdi birkaç tane matematiksel örnek çözelim.

`````python
#bir koordinatın tanımlanması
nokta = (3, 4)
(x, y) = nokta
print("x =", x, "y =", y)
import math as m
uzunluk = m.sqrt(x**2 + y**2)
print("Uzunluk =", uzunluk)
`````

`````python
#iki nokta arası uzaklık
import math as m

A = (3, 4)
B = (7, 1)

(x1, y1) = A
(x2, y2) = B

d = m.sqrt((x2 - x1)**2 + (y2 - y1)**2)
print("Uzaklık =", round(d, 2))
`````

`````python
#üçgenin alanı (paketleme ve açma)
kenarlar = (3, 4, 5)
(a, b, c) = kenarlar

u = (a + b + c) / 2  # yarı çevre
alan = (u*(u-a)*(u-b)*(u-c))**0.5
print("Üçgenin Alanı =", alan)
`````

`````python
#açıların sin ve cos değerleri
import math as m

acilar = (30, 45, 60)
for a in acilar:
    radyan = m.radians(a)
    print(f"{a}° → sin = {round(m.sin(radyan),2)}, cos = {round(m.cos(radyan),2)}")
`````

### 📚 **3.8 Python Sözlükleri (Dictionary)**

**Sözlük (dictionary)**, Python’da **anahtar–değer (key–value)** çiftlerini saklayan veri yapısıdır.
Liste ve tuple’dan farkı, verilere **indeks yerine anahtar** (isim, etiket vb.) ile ulaşılmasıdır.

---

#### 🔹 1. **Sözlük Tanımlama**



`````python
ogrenci = {
    "ad": "Ali",
    "yas": 21,
    "bolum": "Bilgisayar Mühendisliği"
}
`````

#### 🔹 2. **Veriye Erişim**
Sözlük içindeki verilere erişim anahtarlar ile yapılabilir.


`````python
print(ogrenci["ad"])
print(ogrenci.get("yas"))
`````


> 🔸 `get()` metodu, anahtar yoksa hata vermez; `None` döner.


#### 🔹 3. **Yeni Eleman Ekleme ve Güncelleme**
Sözlüklere yeni elemanlar eklenebilir.

`````python
ogrenci["not"] = 90        # Yeni anahtar ekleme
ogrenci["yas"] = 22        # Var olan değeri güncelleme
`````

 #### 🔹 4. **Eleman Silme**

Sözlükten eleman silme ``del`` metodu ile yapılır.

`````python
del ogrenci["not"]          # Belirli anahtarı siler
ogrenci.clear()             # Tüm sözlüğü temizler
`````

#### 🔹 5. **Döngü ile Sözlük Üzerinde Gezme**

Sözlük elemanları üzerinde döngüler çalıştırılabilir.

`````python
notlar = {"Ali": 85, "Ayşe": 92, "Mehmet": 78}

for isim, notu in notlar.items():
    print(isim, "→", notu)
for isim in notlar.keys():
    print(isim)
`````

#### 🔹 6. **Faydalı Metotlar**

| ----------- | ------------------------------------------- |
| `.keys()`   | Tüm anahtarları döner                       |

| `.items()`  | Anahtar–değer çiftlerini döner              |
| `.get(key)` | Anahtarın değerini döner                    |

#### 🔹 7. **İç İçe Sözlük (Nested Dictionary)**

Liste ve tuple da olduğu gibi sözlükler de iç içe yazılabilir.

`````python
ogrenciler = {
    "1001": {"ad": "Ali", "not": 85},
    "1002": {"ad": "Ayşe", "not": 90}
}
print(ogrenciler["1002"]["ad"])  # Ayşe
`````

`````python
#matematiksel bir örnek
acilar = {"α": 30, "β": 45, "γ": 60}
for ad, derece in acilar.items():
    radyan = m.radians(derece)
    print(f"{ad}: sin = {round(m.sin(radyan),2)}  cos = {round(m.cos(radyan),2)}")
`````

### 🔍 **Liste – Tuple – Sözlük (Dictionary) Karşılaştırması**

Python’da **liste**, **tuple** ve **sözlük**, verileri bir arada tutmak için kullanılan üç temel veri yapısıdır.
Ancak aralarında hem **kullanım amacı** hem de **özellik** bakımından farklar vardır:

---

| Özellik                 | **Liste (list)**      | **Tuple (tuple)**     | **Sözlük (dict)**                  |
| ----------------------- | --------------------- | --------------------- | ---------------------------------- |
| **Sözdizimi**           | `[]`                  | `()`                  | `{}`                               |
| **Değiştirilebilirlik** | ✅ Evet (mutable)      | ❌ Hayır (immutable)   | ✅ Evet (mutable)                   |
| **Erişim Şekli**        | İndeks ile            | İndeks ile            | Anahtar (key) ile                  |
| **Eleman Türü**         | Her tür veri olabilir | Her tür veri olabilir | Anahtar–değer (key–value) çiftleri |
| **Kullanım Amacı**      | Sıralı veriler        | Sabit veriler         | Eşleştirilmiş bilgiler             |
| **Örnek**               | `[10, 20, 30]`        | `(10, 20, 30)`        | `{"ad": "Ali", "yas": 21}`         |

---

📘 **Kısaca:**

* **Liste**: Esnek, en sık kullanılan yapıdır.
* **Tuple**: Değişmeyecek sabit veriler için güvenlidir.
* **Sözlük**: Anlamlı eşleştirmeler (ör. ad–not, ürün–fiyat) için idealdir.

## **3.10 Python’da `if` Yapısı**

`if` yapısı, bir koşulun **doğru (True)** olup olmadığını kontrol eder.
Koşul doğruysa ilgili kod bloğu çalışır, yanlışsa atlanır.

---



#### **1. Temel Kullanım**
if deyimi, belirli bir koşul doğru olduğunda kod bloğunu çalıştırmak için kullanılır.
python
```python
if koşul:
    # koşul doğruysa çalışacak kodlar
    ````

`````python
x = 10
if x > 5:
    print("x, 5'ten büyüktür.")
`````

`````python
yaş = int(input("Yaşınızı girin: "))
if yaş >= 18:
    print("Ehliyet alabilirsiniz!")
`````

🔹 **Mantık**

* Koşul doğruysa (`True`) altındaki kod çalışır.
* Yanlışsa (`False`) bu blok **atlanır**.
* Python’da girinti (indentation) çok önemlidir!

---

#### **2. If-else Yapısı**

Koşul doğru değilse alternatif bir kod bloğu çalıştırmak için kullanılır.
python
```python
if koşul:
    # koşul doğruysa çalışacak kodlar
else:
    # koşul yanlışsa çalışacak kodlar
    ```

`````python
if yaş >= 18:
    print("Ehliyet alabilirsiniz!")
else:
    print("Ehliyet almak için yaşınız yeterli değil!")
`````


Birden fazla koşulu kontrol etmek için kullanılır.
```python
if koşul1:
    # koşul1 doğruysa çalışacak kodlar
elif koşul2:
    # koşul2 doğruysa çalışacak kodlar
elif koşul3:
    # koşul3 doğruysa çalışacak kodlar
else:
    # hiçbir koşul doğru değilse çalışacak kodlar
    ```

`````python
notu = 85

if notu >= 90:
    print("Harf Notu: AA")
elif notu >= 80:
    print("Harf Notu: BA")
elif notu >= 70:
    print("Harf Notu: BB")
elif notu >= 60:
    print("Harf Notu: CB")
else:
    print("Harf Notu: FF - Kaldınız!")
`````
