## 6. Modüller Oluşturma


Python’da **modül**, bir dosyada tanımlanmış **fonksiyonlar, değişkenler ve sınıfların** başka programlarda kullanılmasını sağlayan bir yapıdır.
Yani modül, kodu parçalara ayırarak **daha düzenli**, **yeniden kullanılabilir** hale getirir.

Bir modül aslında `.py` uzantılı **normal bir Python dosyasıdır**.

---

### 🔹 **Hazır Modüller**

Python ile birlikte gelen modülleri doğrudan `import` ifadesiyle kullanabiliriz:

```python
import math

print(math.sqrt(25))   # 5.0
print(math.pi)         # 3.141592653589793
```

---

### 🔹 **Belirli Elemanları İçeri Aktarma**

```python

print(sqrt(49))
print(pi)
```

---

### 🔹 **Modülü Kısaltmayla Kullanma**

```python

print(np.arange(5))
```

---


### 🔹 **Sadece Bir Fonksiyonu Almak**

```python

```

---

## 🧩 **6.3. PyPI Nedir?**

**PyPI (Python Package Index)**, Python dünyasında geliştirilen **açık kaynaklı paketlerin ve modüllerin** yayımlandığı merkezi bir **depo (repository)**’dur.
👉 [https://pypi.org](https://pypi.org)

Burada barındırılan binlerce paket, `pip` komutu aracılığıyla kolayca yüklenebilir.
Örneğin:

```bash
pip install numpy
pip install pandas
pip install matplotlib
```

Bu komutlar, ilgili kütüphaneleri **PyPI’den indirip sistemine kurar.**

---

## ⚙️ **Pip Nedir?**

`pip`, “**Pip Installs Packages**” ifadesinin kısaltmasıdır.
Python’un resmi **paket yöneticisidir.**
Yani PyPI’deki bir paketi sistemine indirip, kurulumunu ve bağımlılıklarını yönetir.

Temel komutlar:

| Komut                      | Açıklama                   |
| -------------------------- | -------------------------- |
| `pip install paket_adı`    | Paketi indirip kurar       |
| `pip uninstall paket_adı`  | Paketi kaldırır            |
| `pip show paket_adı`       | Paket bilgilerini gösterir |
| `pip list`                 | Kurulu paketleri listeler  |
| `pip install -U paket_adı` | Paketi günceller           |

---

## 🧠 **PyPI’ye Paket Yükleme Mantığı**

Sen de kendi modülünü (`hesap.py` gibi) geliştirip, PyPI’de paylaşabilirsin.
Böylece herkes `pip install hesap` diyerek senin modülünü yükleyebilir.
Bunun için aşağıdaki adımlar izlenir:

### 🔹 1. Proje dizinini hazırla

```
hesap/
 ├── hesap/
 │   ├── __init__.py
 │   └── islem.py
 ├── setup.py
 ├── README.md
 └── LICENSE
```

### 🔹 2. `setup.py` örneği

```python
from setuptools import setup, find_packages

setup(
    name='hesap',
    version='0.1',
    packages=find_packages(),
    install_requires=[],
    description='Basit matematiksel işlemler modülü',
    author='İnan Ünal',
    author_email='inanunal@example.com',
    url='https://github.com/inanunalml/hesap',
)
```

### 🔹 3. Yükleme komutları

```bash
pip install build twine
python -m build
python -m twine upload dist/*
```

Bu adımlar sonunda senin paketini herkes `pip install hesap` komutuyla indirebilir hale gelir. 🚀

## 🌍 **PyPI’nin Önemi**

* Python ekosisteminin **en büyük paylaşım merkezi**dir.
* Bilimsel kütüphanelerden (NumPy, SciPy) web çerçevelerine (Flask, Django) kadar her şey orada bulunur.
* Açık kaynak topluluklarını destekler.
* Akademik dünyada da yaygın olarak kullanılır (örneğin senin SPD manifold çalışmalarında scikit-learn, geomstats, numpy, scipy gibi paketler hep PyPI kaynaklıdır).

# 🧩 6.4. Python Paketleri, Sürümleme ve Dokümantasyon Güncellemeleri

## 1. Paket Kavramı

**Paket (package)**, birden fazla Python modülünü **düzenli bir yapı içinde birleştiren** dizindir.
Paketler, projelerin büyüdüğü ortamlarda kodun **yeniden kullanılabilirliğini** ve **bakım kolaylığını** sağlar.

Bir paketin en basit biçimi şu şekildedir:

```
hesap/
 ├── __init__.py
 ├── islem.py
 └── istatistik.py
```

* `__init__.py` dosyası, klasörün **bir Python paketi olduğunu belirtir.**
* Diğer `.py` dosyaları bu paketin modülleridir.

Kullanım örneği:

```python
from hesap.islem import topla
```

---

## 2. Paketlerin Kurulumu ve Yönetimi

Python paketleri genellikle **PyPI (Python Package Index)** üzerinden paylaşılır.
`pip` komutu, bu paketleri **indirip sisteminize kurmanızı** sağlar.

### Örnek:

```bash
pip install numpy
pip install matplotlib
```

Kurulu paketleri görmek:

```bash
pip list
```

Belirli bir paketi kaldırmak:

```bash
pip uninstall paket_adi
```

---

## 3. Paket Geliştirme Süreci

Bir Python paketi geliştirirken şu adımlar izlenir:

1. **Klasör yapısı oluşturulur.**
2. **`setup.py` veya `pyproject.toml`** dosyası yazılır.
3. **Bağımlılıklar** (dependencies) belirtilir.
4. **README.md** ve **LICENSE** dosyaları eklenir.
5. **`build`** ve **`twine`** araçlarıyla paket hazırlanır.
6. PyPI’ye yüklenir:

   ```bash
   python -m build
   python -m twine upload dist/*
   ```

Bu işlem sonucunda paket, `pip install hesap` gibi bir komutla erişilebilir hale gelir.

---

## 4. Sürüm Yönetimi (Versioning)

Paketlerin sürdürülebilir olması için **sürümleme (versioning)** önemlidir.
Genellikle **Semantik Sürümleme (Semantic Versioning)** kuralı uygulanır:

```
MAJOR.MINOR.PATCH
```

| Kısım     | Anlamı                                          |
| --------- | ----------------------------------------------- |
| **MAJOR** | Geriye dönük uyumluluğu bozan değişiklikler     |
| **MINOR** | Yeni özellikler eklenmiş ama uyumluluk korunmuş |
| **PATCH** | Küçük hata düzeltmeleri                         |

### Örnek:

* `1.0.0` → İlk stabil sürüm
* `1.1.0` → Yeni fonksiyonlar eklendi
* `1.1.1` → Küçük hata düzeltmesi
* `2.0.0` → Eski fonksiyon yapısı değişti, geriye uyum yok

---

## 5. Dokümantasyonun Önemi

Her paketin kullanıcı dostu olması için **dokümantasyon** zorunludur.
Dokümantasyon şunları içermelidir:

* Kurulum talimatları
* Fonksiyon açıklamaları ve örnek kullanımlar
* Versiyon geçmişi (CHANGELOG)
* Lisans bilgisi
* Katkı rehberi (CONTRIBUTING.md)

Dokümantasyon için yaygın araçlar:

* **Sphinx** → Resmî Python dokümantasyon sistemi
* **MkDocs** → Markdown tabanlı, modern arayüzlü dokümantasyon
* **Jupyter Notebook / Colab** → Etkileşimli örneklerle desteklenen anlatımlar

---

## 6. Dokümantasyon Güncelleme Süreci

Yeni bir sürüm çıktığında dokümantasyonun da güncellenmesi gerekir:

1. **Yeni fonksiyonların açıklamaları eklenir.**
2. **Kullanımdan kaldırılan (deprecated)** kısımlar not edilir.
3. **Versiyon numarası** ve **tarih** güncellenir.
4. **Değişiklikler CHANGELOG.md** dosyasına kaydedilir.
5. Doküman yeniden derlenip web sitesine veya PyPI sayfasına yüklenir.

---
