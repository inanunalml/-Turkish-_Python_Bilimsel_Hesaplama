Numpy'de birçok metot entegre olarak bulunur.

Şimdi Lineer Cebir den bildiğimiz determinant hesabı yapalım. Bunun için Numpy kütüphanesinde bulunan ``linalg`` modülü kullanacağız.

````python
#determinant hesabı
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("A @ B =")
print(A @ B)

detA = np.linalg.det(A)
print("Determinant (A):", detA)
````

``linalg`` modülü içinde bir çok metot bulunur. Aşağıda bu modül ile denklem sistemi çözümü yapılmıştır.

````python
#1.dereceden 2 bilinmeyenli denklem sistemi çözümü
A = np.array([[2, 1],
              [1, -1]])
b = np.array([5, 1])

x = np.linalg.solve(A, b)# A katsayılar matrisi, b sağdaki kolon matrisi
print("Çözüm (x, y) =", x)
````

Şimdi sayısal türev örneği çözelim.

Türev tanımını hatırlayalım:

$$f'(x)= lim_{h→0}\frac{f(x + h) - f(x - h)}{h}$$

Sayısal türevde bu limit yerine belirli bir h değeri ile yaklaşım yapılır:

$$f'(x)≈\frac{f(x + h) - f(x - h)}{2h}$$.
Şimdi bunu kodlayalım.

````python
#sayısal türev
def f(x):
    return np.sin(x)

x = np.pi / 4
h = 1e-5 # bizim seçimimize göre sonuç gerçek değere yaklaşır.
turev = (f(x + h) - f(x - h)) / (2 * h)
print("f'(π/4) ≈", turev)
````

### **7.2 Matplotlib — Bilimsel Verilerin Görselleştirilmesi**

**Matplotlib**, Python’da veri görselleştirmek için en yaygın kullanılan **grafik çizim paketidir.**
NumPy dizileriyle doğrudan çalışabilir ve çizgi grafikleri, çubuk grafikleri, histogramlar, ısı haritaları ve 3D yüzey grafikleri gibi birçok görselleştirme türünü destekler.

En sık kullanılan alt modülü **`pyplot`**’tur ve genellikle şu şekilde içe aktarılır:

````python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)

plt.plot(x, y)
plt.title("Sinüs Fonksiyonu")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.grid(True)
plt.show()
````

## **Kesikli ve Parçalı Fonksiyonları Çizme**

## **Bir Grafikten Animasyon Oluşturma**

Matplotlib'in **FuncAnimation** sınıfı, dinamik grafikler oluşturmak için kullanılır.

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

plt.rcParams["animation.html"] = "jshtml"

fig, ax = plt.subplots()
x = np.linspace(0, 4, 100)
(line,) = ax.plot(x, np.exp(-x) * np.sin(2 * np.pi * x))
ax.set_ylim(-1, 1)

def update(frame):
    line.set_ydata(np.exp(-x) * np.sin(2 * np.pi * (x + frame / 10)))
    return (line,)

ani = animation.FuncAnimation(fig, update, frames=100, interval=50, blit=True)
HTML(ani.to_jshtml())   # Colab'ta böyle göm
````

### **7.3 SciPy — Bilimsel Hesaplama Paketleri**

**SciPy (Scientific Python)**, NumPy üzerine inşa edilmiş bir **bilimsel hesaplama paketidir.**
İstatistik, optimizasyon, integral alma, türev hesaplama, lineer cebir, sinyal işleme ve diferansiyel denklemlerin çözümü gibi konularda zengin araçlar sunar.

SciPy’nin gücü, karmaşık matematiksel işlemleri **yalnızca birkaç satır kod** ile gerçekleştirebilmesinden gelir.
Paket, farklı alt modüllere ayrılmıştır:

| Alt Modül         | Açıklama                             |
| ----------------- | ------------------------------------ |
| `scipy.integrate` | Sayısal integral hesaplamaları       |
| `scipy.optimize`  | Optimizasyon, kök bulma işlemleri    |
| `scipy.linalg`    | Lineer cebir (matris) işlemleri      |
| `scipy.stats`     | Olasılık ve istatistik fonksiyonları |
| `scipy.fft`       | Fourier dönüşümleri                  |
| `scipy.signal`    | Sinyal işleme araçları               |

---

### ⚙️ **Örnek: Kök Bulma ve İntegral Hesaplama**


````python
import numpy as np
from scipy import optimize, integrate

# Örnek fonksiyon: f(x) = cos(x) - x
f = lambda x: np.cos(x) - x

# 1️⃣ Kök bulma (f(x)=0 için çözüm)
kok = optimize.root_scalar(f, bracket=[0, 1], method='brentq')
print("Kök:", kok.root)

# 2️⃣ İntegral hesaplama (0–π arası sin(x))
sonuc, hata = integrate.quad(np.sin, 0, np.pi)
print("∫ sin(x) dx =", sonuc)
````

### 🧠 **Kod Açıklaması**

* `optimize.root_scalar()` → Verilen aralıkta fonksiyonun kökünü bulur.
* `integrate.quad()` → Sayısal integral hesaplar.
* `brentq` → Kök bulmak için Brent algoritmasını kullanır (hızlı ve kararlı).
---

Bu örnek SciPy’nin iki temel gücünü gösterir:

1. **Optimizasyon:** karmaşık fonksiyonların kök ve minimum noktalarını bulabilir.
2. **İntegrasyon:** analitik çözümü olmayan fonksiyonlarda yüksek doğrulukla alan hesabı yapabilir.

SciPy, mühendislik ve bilimsel araştırmalarda **NumPy’nin doğal tamamlayıcısıdır.**

### **7.4 SymPy — Sembolik Matematik**

**SymPy (Symbolic Python)**, Python’da **sembolik matematik** yapmak için geliştirilmiş bir pakettir.
Yani sayısal sonuçlar yerine **matematiksel ifadeler** üzerinde işlem yapar.
Türev, integral, denklem çözümü, limit, matris işlemleri ve hatta LaTeX çıktısı üretimi gibi işlemleri destekler.

SymPy, bilimsel hesaplamada **analitik düşünmeyi** kolaylaştırır; öğrencilerin bir fonksiyonun davranışını adım adım görmesini sağlar.

---

### ⚙️ **Sembolik Hesaplamaya Giriş**
 Şimdi $f(x)=sinx+x^2$ fonksiyonunu inceleyelim.

---


````python
import sympy as sp

# Sembolik değişken tanımlama
x = sp.Symbol('x')

# Fonksiyon tanımı
f = sp.sin(x) + x**2

# Türev
turev = sp.diff(f, x)

# İntegral
integral = sp.integrate(f, x)

# Limit
limit = sp.limit(f/x, x, 0)


print("f(x) =", f)
print("f'(x) =", turev)
print("∫f(x)dx =", integral)
print("lim f(x)/x =", limit)
````

### 🧠 **Kod Açıklaması**

* `sp.Symbol('x')` → `x` değişkenini sembolik hale getirir.
* `sp.diff()` → Analitik türev alır.
* `sp.integrate()` → Belirsiz veya belirli integral hesaplar.
* `sp.limit()` → Limit hesaplaması yapar.

Sembolik olarak verilen denklemlerin köklerini bulabiliriz.

Bunun için
* `sp.solve()` metodu kullanılır.

````python
import sympy as sp

# Sembolik değişken tanımlama
x = sp.Symbol('x')
f=3*x**2-2*x
# 4️⃣ Denklem çözümü
denklem = sp.solve(sp.Eq(f, 1), x)
print("f(x)=1 denkleminin kökleri:", denklem)
````

### 💡 **Sonuç**

SymPy, matematiksel işlemleri yalnızca hesaplamakla kalmaz, **adım adım sadeleştirir** ve sonuçları `LaTeX` biçiminde ifade edebilir:

```python
sp.latex(integral)
```

çıktısı, doğrudan bilimsel rapor veya makalelere eklenebilecek biçimdedir.

SymPy, özellikle **analitik düşünme, model türetme ve eğitim amaçlı** çalışmalarda son derece yararlıdır.

````python
import sympy as sp
from IPython.display import display, Math

# Sembolik değişken ve fonksiyon
x = sp.Symbol('x')
f = sp.sin(x) + x**2

# Türev ve integral
df = sp.diff(f, x)
I = sp.integrate(f, x)

# LaTeX biçiminde görüntüleme
display(Math(r"f(x) = " + sp.latex(f)))
display(Math(r"f'(x) = " + sp.latex(df)))
display(Math(r"\int f(x)\,dx = " + sp.latex(I)))
````

````python
#örnek matrisler
A = sp.Matrix([[2, 1], [1, 3]])
B = sp.Matrix([[1, 0], [4, 2]])

# Matris çarpımı, determinant, ters
carpim = A * B
detA = A.det()
invA = A.inv()

sp.pretty_print(carpim)
print("det(A) =", detA)
sp.pretty_print(invA)
````

````python
from IPython.display import display, Math
x = sp.Symbol('x')
f = sp.exp(-x**2)

df = sp.diff(f, x)
I = sp.integrate(f, (x, -sp.oo, sp.oo))

display(Math("f(x) = " + sp.latex(f)))
display(Math("f'(x) = " + sp.latex(df)))
display(Math(r"\int_{-\infty}^{\infty} e^{-x^2}\,dx = " + sp.latex(I)))
````

````python
#örnek, çok değişkenli fonksiyonlarda türev
x, y = sp.symbols('x y')
f = x**2 * sp.exp(y)

# Kısmi türevler
dfdx = sp.diff(f, x)
dfdy = sp.diff(f, y)

# İkinci türevler (Hessian matrisi)
H = sp.hessian(f, (x, y))

display(Math(r"\text{Kısmi türev: } \frac{\partial f}{\partial x} = " + sp.latex(dfdx)))

display(Math(r"\text{Kısmi türev: } \frac{\partial f}{\partial y} = " + sp.latex(dfdy)))

display(Math(r"\text{Hessian: } H = " + sp.latex(H)))
````
