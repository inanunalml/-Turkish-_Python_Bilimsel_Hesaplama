## **UYARI: eval ve exec**

## eval() (Değerlendirme)
- **String ifadeleri değerlendirir** ve sonuç döndürür
- **Matematiksel işlemler** ve basit ifadeler için kullanılır
- **Tek bir ifade** ile çalışır

### Örnek:
```python
sonuc = eval("5 + 3 * 2")
print(sonuc)  # 11
```

## exec() (Çalıştırma)
- **String kod bloklarını çalıştırır**
- **Birden fazla satır** ve karmaşık işlemler için kullanılır
- **Sonuç döndürmez** (None döndürür)

### Örnek:
```python
exec("""
for i in range(3):
    print(f'Döngü: {i}')
""")
```

Veriyi bir liste gibi düşünerek onun üzerinden işlemler yapabiliriz.

````python
 #Dosyayı aç ve oku
with open(veri, 'r') as f:
    veriler = f.readlines()

# Her satırı işle
for v in veriler:
    v = v.strip()
    # Yorum satırlarını (# ile başlayanları) atla
    if v.startswith('#') or v == '':
        continue
    # Sıcaklık ve Gerilim değerlerini ayır
    sicaklik, gerilim = v.split()
    print(f"Sıcaklık: {float(sicaklik):.2f} °C,  Gerilim: {float(gerilim):.2f} mV")
````

## **5.5 Programlarda Hata Yönetimi (try-except)**
Python’da hata yönetimi (error handling), program çalışırken ortaya çıkabilecek hataları yakalayıp, programın çökmeden devam etmesini sağlamaktır. Bunun için en sık kullanılan yapı `try-except` bloklarıdır.

````python
try:
    yas = int(input('Yaşınızı girin: '))
    print(f'Girdiğiniz yaş: {yas}')
except ValueError:
    print('Lütfen geçerli bir sayı girin.')
````

Örneğin sıfır bölmeyi önleyebiliriz.

````python
try:
    sayı = int(input("Bir sayı girin: "))
    print(10 / sayı)
except:
    print("Bir hata oluştu!")
````

Bazı özel kod ifadeleri ile belirli hataları yakalayabiliriz. Örneğin sıfır bölme hatası için:


````python
try:
    sayı = int(input("Bir sayı girin: "))
    print(10 / sayı)
except ValueError:
    print("Lütfen geçerli bir sayı girin.")
except ZeroDivisionError:
    print("Bir sayıyı sıfıra bölemezsiniz!")
````

Eğer hata yoksa çalışmasını isteyebilir, veya hata varsa da programın çalışmasını isteyebiliriz.

````python
try:
    sayı = int(input("Bir sayı girin: "))
    sonuç = 10 / sayı
except ZeroDivisionError:
    print("Sıfıra bölme hatası!")
else:
    print(f"Sonuç: {sonuç}")
finally:
    print("Program sonlandı.")
````

``else``: Hata oluşmazsa çalışır.

``finally``: Her durumda (hata olsa da olmasa da) çalışır.

Bir fonksiyon yazımındaki muhtemel hataları çıkarabiliriz.

````python
def karekök(x):
    if x < 0:
        raise ValueError("Negatif sayıların karekökü alınamaz.")
    return x ** 0.5

#kodu çalıştırıp hatayı görelim
````

````python
karekök(-2)
````

````python
try:
    print(karekök(-5))
except ValueError as e:
    print("Hata:", e)
# Programın çıktı vermesini sağladık.
````

🔍 **Özetle**

| Yapı      | Açıklama                       |
| --------- | ------------------------------ |
| `try`     | Hata oluşabilecek kod bloğu    |
| `except`  | Hata yakalanınca çalışan kısım |
| `else`    | Hata oluşmazsa çalışan kısım   |
| `finally` | Her durumda çalışan kısım      |
| `raise`   | Manuel olarak hata oluşturma   |
